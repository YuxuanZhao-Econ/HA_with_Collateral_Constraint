# Heterogeneous Agents in a Small Open Economy with Collateral Borrowing Constraints

- Authors: Yuxuan Zhao
- Date: 2026-09-23

This notebook defines a heterogeneous-household extension of the endowment economy in [Bianchi (2011)](../reference/2011-overborrowing-and-systemic-externalities-in-the-business-cycle.pdf)

Households differ in tradable endowments and bond holdings. The aggregate shock follows a Markov chain; idiosyncratic income shocks are IID. The full asset distribution is part of the aggregate state, and the joint asset-income distribution factors into the asset distribution and the fixed IID income distribution. The exposition proceeds from economic fundamentals to the recursive household problem and then to the recursive competitive equilibrium. All equilibrium objects are defined using the full distribution.

## 1. Economic fundamentals

### 1.1 Households, goods, and financial markets

Time is discrete, $t=0,1,2,\ldots$. There is a unit mass of infinitely lived, price-taking households indexed by $i\in[0,1]$. Households have common preferences and face uninsurable idiosyncratic income risk as well as aggregate risk.

There are two consumption goods. Tradables can be exchanged internationally and their price is normalized to one. Nontradables are exchanged within the domestic economy at the common relative price $p_t^N>0$.

Each household can trade a one-period, non-state-contingent bond with the rest of the world at the exogenous gross return
$$
R=1+r>1.
$$
The bond is denominated in tradable goods. The inherited position is $b_{i,t}$ and the position chosen at date $t$ is $b_{i,t+1}$. Positive holdings represent assets; negative holdings represent debt. Households cannot trade state-contingent claims. Both sectors are endowment sectors.

### 1.2 Aggregate and idiosyncratic endowments

Let $z_t\in\mathcal Z$ be a finite-state aggregate Markov shock, with
$$
\Pr(z_{t+1}=z'\mid z_t=z)=\Pi(z'\mid z),
\qquad \sum_{z'\in\mathcal Z}\Pi(z'\mid z)=1.
$$
The common endowment components are functions $y_c^T(z)$ and $y^N(z)>0$. Household $i$ has an idiosyncratic tradable-income component $\varepsilon_{i,t}\in\mathcal E$, where $\mathcal E$ is a finite set. Its endowments are
$$
\boxed{
y_{i,t}^T=y^T(z_t,\varepsilon_{i,t})
=y_c^T(z_t)+\varepsilon_{i,t},
\qquad y_{i,t}^N=y^N(z_t).
}
$$
Choose the endowment functions and shock supports so that
$$
y^T(z,\varepsilon)>0
\qquad\text{for all admissible }(z,\varepsilon).
$$

Idiosyncratic shocks are independent and identically distributed across households and time, and independent of the entire aggregate shock process:
$$
\boxed{\varepsilon_{i,t}\overset{\mathrm{iid}}{\sim}\pi_\varepsilon,}
\qquad
\pi_\varepsilon(\varepsilon)\geq0,\quad
\sum_{\varepsilon\in\mathcal E}\pi_\varepsilon(\varepsilon)=1,\quad
\sum_{\varepsilon\in\mathcal E}\varepsilon\pi_\varepsilon(\varepsilon)=0.
$$
Thus the household's joint transition probabilities are
$$
\Pr(z',\varepsilon'\mid z,\varepsilon)
=\Pi(z'\mid z)\pi_\varepsilon(\varepsilon').
$$
Current and past individual income shocks do not predict the next idiosyncratic draw. IID shocks still allow runs of favorable or adverse realizations; past shocks can have lasting effects through accumulated assets and debt.

Assume a conditional exact law of large numbers for the continuum of households. Conditional on the aggregate history, the idiosyncratic cross-sectional distribution is $\pi_\varepsilon$, including within groups defined by inherited assets. Initial assets are independent of the initial idiosyncratic draw. Consequently,
$$
\int_0^1\varepsilon_{i,t}\,di=0,
\qquad
Y_t^T=\int_0^1 y_{i,t}^T\,di=y_c^T(z_t).
$$

### 1.3 Preferences

Each household maximizes
$$
\mathbb E_0\sum_{t=0}^{\infty}\beta^t u(C_{i,t}),
\qquad 0<\beta<1,
$$
where
$$
u(C)=
\begin{cases}
\dfrac{C^{1-\sigma}}{1-\sigma}, & \sigma>0,\ \sigma\ne1,\\
\log C, & \sigma=1,
\end{cases}
$$
and the consumption basket is
$$
C_{i,t}=C(c_{i,t}^T,c_{i,t}^N)
=\left[\omega(c_{i,t}^T)^{-\eta}
+(1-\omega)(c_{i,t}^N)^{-\eta}\right]^{-1/\eta},
\qquad 0<\omega<1,\quad\eta>-1,\quad\eta\ne0.
$$
At $\eta=0$, use the continuous Cobb-Douglas limit
$$
C(c^T,c^N)=(c^T)^\omega(c^N)^{1-\omega}.
$$
The elasticity of substitution between tradable and nontradable consumption is $1/(1+\eta)$. Both consumption goods are strictly positive.

### 1.4 Budget and collateral borrowing constraints

For every household and date,
$$
\boxed{
b_{i,t+1}+c_{i,t}^T+p_t^Nc_{i,t}^N
=Rb_{i,t}+y^T(z_t,\varepsilon_{i,t})+p_t^Ny^N(z_t).
}
$$
Borrowing is limited by the market value of the household's current endowments:
$$
\boxed{
b_{i,t+1}\geq
-\left[\kappa^Ty^T(z_t,\varepsilon_{i,t})
+\kappa^Np_t^Ny^N(z_t)\right],
\qquad 0\leq\kappa^T,\kappa^N<1.
}
$$
The parameters $\kappa^T$ and $\kappa^N$ are pledgeable fractions. A negative idiosyncratic income shock reduces both current resources and, when $\kappa^T>0$, borrowing capacity. A lower common price $p_t^N$ reduces every household's nontradable collateral value when $\kappa^N>0$.

Households take prices as given. They may buy or sell nontradables domestically, so individual consumption need not equal individual endowment:
$$
c_{i,t}^N\ne y^N(z_t)\quad\text{in general}.
$$
Admissible plans are adapted to observed histories, satisfy both constraints with positive consumption at every history of positive probability, have well-defined expected lifetime utility, and obey the no-Ponzi restriction
$$
\liminf_{T\to\infty}R^{-T}b_{i,T}\geq0
\qquad\text{almost surely}.
$$
There is no default or additional transfer instrument. Equilibrium is defined on states admitting feasible continuation plans.

## 2. States, distribution, and timing

### 2.1 Individual and aggregate states

Let $F_t$ be the cross-sectional probability distribution of inherited bond holdings and let $\Phi_t$ denote the joint distribution of inherited assets and current idiosyncratic income:
$$
F_t(A)=\int_0^1\mathbf 1\{b_{i,t}\in A\}\,di,
\qquad
\Phi_t(A\times E)=\int_0^1
\mathbf 1\{b_{i,t}\in A,\ \varepsilon_{i,t}\in E\}\,di,
$$
for measurable $A\subseteq\mathbb R$ and $E\subseteq\mathcal E$.

Assets $b_{i,t}$ are chosen before the fresh IID draw $\varepsilon_{i,t}$. Conditional on the aggregate history, their joint distribution therefore factors as
$$
\boxed{
\Phi_t(A\times E)=F_t(A)\sum_{\varepsilon\in E}\pi_\varepsilon(\varepsilon),
\qquad \Phi_t=F_t\otimes\pi_\varepsilon.
}
$$
It is sufficient to track the full asset distribution $F_t$, with
$$
\int_{\mathbb R}dF_t(b)=1,
\qquad \int_{\mathbb R}|b|\,dF_t(b)<\infty.
$$
The household's own state is $(b,\varepsilon)$, the aggregate state is $(z,F)$, and the full decision state is
$$
\boxed{s=(b,\varepsilon,z,F).}
$$
Current $\varepsilon$ remains a household state because it affects current resources and the collateral limit. Inherited assets retain the effects of past shocks. Assets chosen today, $b_{i,t+1}$, generally depend on today's $\varepsilon_{i,t}$; the independence statement concerns $b_{i,t}$ and $\varepsilon_{i,t}$.

Aggregate net foreign assets and endowments are
$$
B(F)=\int b\,dF(b),
\qquad
Y^T(z)=\sum_{\varepsilon\in\mathcal E}\pi_\varepsilon(\varepsilon)
\int y^T(z,\varepsilon)\,dF(b)=y_c^T(z),
\qquad Y^N(z)=y^N(z).
$$
Here and below, asset integrals are over $\mathbb R$. The factorization is exact under the IID assumptions; it does not replace the asset distribution with its mean $B$. Different asset distributions can generate different consumption and borrowing decisions even when their means coincide.

### 2.2 Within-period timing and aggregate laws

1. Households enter date $t$ with inherited assets distributed according to $F_t$. Current shocks $z_t$ and $\varepsilon_{i,t}$ are observed. The joint asset-income distribution is $\Phi_t=F_t\otimes\pi_\varepsilon$.
2. Given the aggregate state, households trade nontradables at $p_t^N=p^N(z_t,F_t)$ and choose $(c_{i,t}^T,c_{i,t}^N,b_{i,t+1})$ simultaneously.
3. These saving choices determine the next inherited-asset distribution $F_{t+1}$. Next period's aggregate shock is drawn from $\Pi$ and fresh idiosyncratic shocks are drawn independently from $\pi_\varepsilon$. The next joint distribution is $\Phi_{t+1}=F_{t+1}\otimes\pi_\varepsilon$.

Given a perceived asset-distribution law $\mathcal H$, households forecast
$$
\boxed{F'=\mathcal H(F,z),}
\qquad p^{N\prime}=p^N(z',F').
$$
The distribution $F'$ is determined before $z'$ is realized, while next-period prices and choices depend on the realized $z'$. Households observe the aggregate state and their own states and know $\Pi$ and $\pi_\varepsilon$. Each household has measure zero, so its own deviation in $b'$ does not change $F'$ or the price functions it faces.

## 3. Recursive household problem

### 3.1 Bellman equation

Given the common price function $p^N$ and perceived law of motion $\mathcal H$, the household solves
$$
\boxed{
\begin{aligned}
V(b,\varepsilon;z,F)
=\max_{b',\,c^T>0,\,c^N>0}\ \Bigg\{&
u\!\left(C(c^T,c^N)\right)\\
&+\beta\sum_{z'\in\mathcal Z}\Pi(z'\mid z)
\sum_{\varepsilon'\in\mathcal E}\pi_\varepsilon(\varepsilon')
V\!\left(b',\varepsilon';z',\mathcal H(F,z)\right)
\Bigg\},
\end{aligned}
}
$$
subject to
$$
\begin{aligned}
b'+c^T+p^N(z,F)c^N
&=Rb+y^T(z,\varepsilon)+p^N(z,F)y^N(z),\\
b'&\geq\underline b(\varepsilon;z,F)
\equiv-\left[\kappa^Ty^T(z,\varepsilon)
+\kappa^Np^N(z,F)y^N(z)\right],\\
F'&=\mathcal H(F,z).
\end{aligned}
$$
The maximization is over choices admitting an admissible continuation plan as defined above. Its policy functions are
$$
b'=g(b,\varepsilon;z,F),\qquad
c^T=c^T(b,\varepsilon;z,F),\qquad
c^N=c^N(b,\varepsilon;z,F).
$$
The borrowing bound depends on the household's current income and the common equilibrium price. The expectation includes the Markov aggregate shock and a fresh IID income draw. The next asset distribution $F'=\mathcal H(F,z)$ is already determined by current decisions; next-period prices also depend on $z'$.

### 3.2 Household optimality conditions

Define marginal utilities with respect to the two goods by
$$
u_T(c^T,c^N)=u'(C(c^T,c^N))C_T(c^T,c^N),
\qquad
u_N(c^T,c^N)=u'(C(c^T,c^N))C_N(c^T,c^N).
$$
Let $\lambda(s)$ be the budget multiplier and $\mu(s)\geq0$ the collateral multiplier. Under the usual differentiability conditions, optimal policies satisfy
$$
\lambda(s)=u_T(c^T(s),c^N(s)),
\qquad
p^N(z,F)\lambda(s)=u_N(c^T(s),c^N(s)).
$$
In particular, optimal intratemporal allocation requires
$$
p^N(z,F)=\frac{1-\omega}{\omega}
\left(\frac{c^T(s)}{c^N(s)}\right)^{1+\eta}.
$$
For a transition $(z',\varepsilon')$, write
$$
s'=\left(g(s),\varepsilon',z',\mathcal H(F,z)\right).
$$
The Euler and complementary-slackness conditions are
$$
\lambda(s)=\beta R
\sum_{z'\in\mathcal Z}\Pi(z'\mid z)
\sum_{\varepsilon'\in\mathcal E}\pi_\varepsilon(\varepsilon')
\lambda(s')+\mu(s),
$$
$$
\begin{aligned}
\Delta(s)&=g(s)+\kappa^Ty^T(z,\varepsilon)
+\kappa^Np^N(z,F)y^N(z)\geq0,\\
\mu(s)&\geq0,\qquad \mu(s)\Delta(s)=0.
\end{aligned}
$$
These are private optimality conditions: households do not differentiate the equilibrium price with respect to their own choices. Lifetime optimality is specified by the household problem, rather than by these local conditions alone.

## 4. Recursive competitive equilibrium

Given the primitives
$$
\left\{\beta,\sigma,\omega,\eta,R,\kappa^T,\kappa^N,
\mathcal Z,\mathcal E,\Pi,\pi_\varepsilon,y_c^T,y^N\right\}
$$
and an admissible initial state $(z_0,F_0)$ with $\Phi_0=F_0\otimes\pi_\varepsilon$, a recursive competitive equilibrium is a collection
$$
\boxed{\left\{V,\ g,\ c^T,\ c^N,\ p^N,\ \mathcal H\right\}}
$$
and its induced distribution process
$$
F_{t+1}=\mathcal H(F_t,z_t),
\qquad z_{t+1}\sim\Pi(\cdot\mid z_t),
$$
satisfying the following conditions at every reachable aggregate state $(z,F)$.

### 4.1 Household optimization

Given $p^N$ and $\mathcal H$, $V$ is the value function of the recursive household problem. For every admissible individual state, the policy triple attains its maximum:
$$
\begin{aligned}
&\left(g(s),c^T(s),c^N(s)\right)\\
&\quad\in\operatorname*{arg\,max}_{b',\,x^T>0,\,x^N>0}
\left\{u(C(x^T,x^N))+
\beta\sum_{z',\varepsilon'}\Pi(z'\mid z)\pi_\varepsilon(\varepsilon')
V(b',\varepsilon';z',\mathcal H(F,z))\right\},
\end{aligned}
$$
subject to the budget, collateral, and continuation-feasibility restrictions in Section 3.1. In particular,
$$
g(s)+c^T(s)+p^N(z,F)c^N(s)
=Rb+y^T(z,\varepsilon)+p^N(z,F)y^N(z),
\qquad g(s)\geq\underline b(\varepsilon;z,F).
$$

### 4.2 Consistency of the distribution law

Given the saving policy $g$ and IID income probabilities $\pi_\varepsilon$, the actual asset-distribution law $\mathcal G_g$ is defined, for every measurable asset set $A\subseteq\mathbb R$, by
$$
\boxed{
\mathcal G_g(F,z)(A)
=\sum_{\varepsilon\in\mathcal E}\pi_\varepsilon(\varepsilon)
\int\mathbf 1\!\left\{g(b,\varepsilon;z,F)\in A\right\}\,dF(b).
}
$$
Rational expectations require the perceived and actual laws to coincide as probability measures:
$$
\boxed{\mathcal H(F,z)=\mathcal G_g(F,z).}
$$
Fresh IID income draws then imply, for every $\varepsilon'\in\mathcal E$,
$$
\Phi'(A\times\{\varepsilon'\})
=\mathcal H(F,z)(A)\pi_\varepsilon(\varepsilon').
$$
The sum in the asset update averages over current shocks $\varepsilon$, which affect today's saving decisions. The next income draw $\varepsilon'$ is independent of the assets already selected. Both distribution statements hold conditional on any next aggregate realization $z'$ with $\Pi(z'\mid z)>0$.

### 4.3 Nontradable goods market clearing

At the common price $p^N(z,F)>0$, aggregate nontradable demand equals the domestic endowment:
$$
\boxed{
\sum_{\varepsilon\in\mathcal E}\pi_\varepsilon(\varepsilon)
\int c^N(b,\varepsilon;z,F)\,dF(b)=y^N(z).
}
$$
Define aggregate consumption of good $j\in\{T,N\}$ by
$$
C^j(z,F)=\sum_{\varepsilon\in\mathcal E}\pi_\varepsilon(\varepsilon)
\int c^j(b,\varepsilon;z,F)\,dF(b).
$$
Here $C^T$ and $C^N$ denote aggregate quantities of the separate goods; $C(c^T,c^N)$ denotes an individual CES basket.

### 4.4 Bond aggregation and the tradable resource constraint

Current and next-period aggregate net foreign assets satisfy
$$
B=B(F)=\int b\,dF(b),
\qquad
B'=\sum_{\varepsilon\in\mathcal E}\pi_\varepsilon(\varepsilon)
\int g(b,\varepsilon;z,F)\,dF(b).
$$
Distribution consistency also implies
$$
B\!\left(\mathcal H(F,z)\right)=B'.
$$
Both $F'$ and its mean $B'$ are determined by choices made before $z'$ is realized.

With mean-zero IID income shocks, the aggregate tradable resource constraint is
$$
\boxed{C^T(z,F)+B'=RB(F)+y_c^T(z).}
$$
This condition follows by aggregating household budgets and imposing nontradable market clearing. Foreign lenders accommodate the net bond position at the world return $R$; domestic net bond holdings are not required to sum to zero.

### 4.5 An implied aggregate price relation

With common CES preferences and a common price, household optimality implies
$$
c^N(b,\varepsilon;z,F)
=q\!\left(p^N(z,F)\right)c^T(b,\varepsilon;z,F),
\qquad
q(p)=\left(\frac{1-\omega}{\omega p}\right)^{1/(1+\eta)}.
$$
Aggregating and using nontradable market clearing gives
$$
q\!\left(p^N(z,F)\right)C^T(z,F)=y^N(z),
$$
or equivalently,
$$
\boxed{
p^N(z,F)=\frac{1-\omega}{\omega}
\left(\frac{C^T(z,F)}{y^N(z)}\right)^{1+\eta}.
}
$$
This is a consequence of household optimality and market clearing, not an additional independent equilibrium condition. It does not impose $c_i^N=y^N$ household by household. The full distribution affects aggregate consumption through individual saving and borrowing decisions, which in turn determine the common relative price and collateral values.

## 6. Numerical algorithm

### 6.1 Approximate aggregate state and grids

Households summarize the distribution by mean assets:
$$
B=\int b\,dF(b),\qquad s=(b,\varepsilon;z,B).
$$
Start from trial aggregate forecasting functions:
$$
\boxed{p^N=\widehat P(z,B),\qquad B'=\widehat H(z,B).}
$$
Next period's perceived price is $\widehat P(z',\widehat H(z,B))$. Current saving determines $B'$ before $z'$ is realized, so $\widehat H$ has no $z'$ argument. Simulation retains the full distribution $F_t$.

| Grid | Domain | Nodes |
| --- | --- | ---: |
| Individual policy assets b | [-1.35, 1.00] | 300 |
| Inherited-asset distribution F (same b nodes) | [-1.35, 1.00] | 300 |
| Mean assets B | [-1.10, -0.30] | 160 |

The shared individual/distribution grid puts 240 nodes in $[-1.35,-0.45]$ and 60 additional nodes above $-0.45$. The $B$ grid has 17 nodes in $[-1.10,-1.02]$, 100 additional nodes through $-0.82$ (spacing $0.002$), and 43 additional nodes through $-0.30$. Use linear interpolation between continuous-state nodes and in log prices. Grid endpoints are numerical truncations, not economic borrowing limits.

### 6.2 Household policies for given aggregate rules

Fix $(\widehat P,\widehat H)$ and let $p=\widehat P(z,B)$. The within-period first-order condition and budget imply
$$
q(p)=\left(\frac{1-\omega}{\omega p}\right)^{1/(1+\eta)},\qquad
c^N=q(p)c^T,\qquad
c^T=\frac{x-b'}{d(p)},
$$
$$
x=Rb+y_c^T(z)+\varepsilon+p y^N(z),\qquad d(p)=1+p q(p).
$$
Write composite consumption and marginal utility as
$$
A_c(p)=[\omega+(1-\omega)q(p)^{-\eta}]^{-1/\eta},\qquad
C=A_c(p)c^T,
$$
$$
\lambda=u_T(c^T,c^N)=K(p)(c^T)^{-\sigma},\qquad
K(p)=\omega A_c(p)^{1+\eta-\sigma}.
$$
Thus both goods remain household choices; only their optimal ratio has been substituted out.

Start from a feasible saving policy $g^{(0)}$. For policy iteration $k$, repeat:

**1. Expected future marginal utility.** From the old policy, calculate its marginal-utility table $\lambda^{(k)}$ and
$$
\mathcal M^{(k)}(a;z,B)
=\beta R\sum_{z',\varepsilon'}\Pi_{zz'}\pi_\varepsilon(\varepsilon')
\lambda^{(k)}\!\left(a,\varepsilon';z',\widehat H(z,B)\right).
$$
Evaluate the table by interpolation and use the old saving choice:
$$
M^{(k)}=\mathcal M^{(k)}\!\left(g^{(k)}(b,\varepsilon;z,B);z,B\right).
$$
Hold this scalar fixed for the current update.

**2. Consumption at the borrowing limit.** Compute
$$
\ell=-\kappa^T[y_c^T(z)+\varepsilon]-\kappa^N p y^N(z),\qquad
c^T_{\rm bound}=\frac{x-\ell}{d(p)}.
$$

**3. Update consumption and saving.** The binding and interior proposals are
$$
\widetilde c^T=
\begin{cases}
c^T_{\rm bound},&K(p)(c^T_{\rm bound})^{-\sigma}\ge M^{(k)},\\[3pt]
\left[K(p)/M^{(k)}\right]^{1/\sigma},&K(p)(c^T_{\rm bound})^{-\sigma}<M^{(k)},
\end{cases}
$$
$$
\widetilde g=x-d(p)\widetilde c^T,\qquad
\widetilde c^N=q(p)\widetilde c^T.
$$
In the first case the household borrows to its limit; in the second it chooses higher saving to satisfy the Euler equality. Apply the numerical asset domain, update the policy table, and repeat until the policy converges. This yields $(g,\lambda)$ for the given aggregate rules.

### 6.3 Current market clearing and distribution transition

At simulated state $(F_t,z_t)$, the distribution nodes coincide with the policy nodes: $b_j^F=b_j$. Compute
$$
B_t=\sum_j b_j^F F_{j,t},\qquad
\mathcal M_t(a)=\beta R\sum_{z',\varepsilon'}\Pi_{z_tz'}\pi_\varepsilon(\varepsilon')
\lambda\!\left(a,\varepsilon';z',\widehat H(z_t,B_t)\right).
$$
Keep this continuation function fixed while searching for today's price. At each trial $p$, repeat the household calculation in Section 6.2 at the occupied asset nodes, using $\mathcal M_t(g)$ on the Euler right-hand side. Denote the resulting choices by $g_t(b,\varepsilon;p)$ and $c_t^N(b,\varepsilon;p)$.

Solve the scalar market-clearing equation
$$
\boxed{\sum_{j,\varepsilon}F_{j,t}\pi_\varepsilon(\varepsilon)
c_t^N(b_j^F,\varepsilon;p_t)=y^N(z_t).}
$$
Changing $p$ changes income, collateral, and current choices. Therefore households must be solved again at each trial price. Future expectations continue to use the perceived rules.

Move the mass associated with saving $g_t$ to its two neighboring distribution nodes. For $b_j^F\le g_t\le b_{j+1}^F$,
$$
w_j(g_t)=\frac{b_{j+1}^F-g_t}{b_{j+1}^F-b_j^F},\qquad
w_{j+1}(g_t)=1-w_j(g_t),
$$
with all other weights zero. The distribution and aggregate updates are
$$
\boxed{F_{k,t+1}=\sum_{j,\varepsilon}F_{j,t}\pi_\varepsilon(\varepsilon)
w_k\!\left(g_t(b_j^F,\varepsilon;p_t)\right),}
$$
$$
B_{t+1}=\sum_k b_k^F F_{k,t+1},\qquad
\Phi_{t+1}=F_{t+1}\otimes\pi_\varepsilon,\qquad
C_t^T+B_{t+1}=RB_t+y_c^T(z_t).
$$
The two-node allocation preserves mass and mean saving. The actual $B_{t+1}$ is carried forward; it need not equal its forecast $\widehat H(z_t,B_t)$.

### 6.4 Update the forecasting functions from the simulated path

For each aggregate state $z$, store the forecasts at the ordered mean-asset nodes $B_1<\cdots<B_J$:
$$
a_{j,z}=\log\widehat P(z,B_j),\qquad h_{j,z}=\widehat H(z,B_j).
$$
For $B_j\le B\le B_{j+1}$, use the line joining the two neighboring values:
$$
\boxed{\begin{aligned}
\log\widehat P(z,B)
&=a_{j,z}+\frac{B-B_j}{B_{j+1}-B_j}(a_{j+1,z}-a_{j,z}),\\
\widehat H(z,B)
&=h_{j,z}+\frac{B-B_j}{B_{j+1}-B_j}(h_{j+1,z}-h_{j,z}).
\end{aligned}}
$$
The initial guess sets $\bar B=J^{-1}\sum_j B_j$ and
$$
\widehat H_0(z,B)=\bar B+0.85(B-\bar B),\qquad
\widehat P_0(z,B)=\frac{1-\omega}{\omega}
\left(\frac{y_c^T(z)+(R-1)B}{y^N(z)}\right)^{1+\eta}.
$$

Simulate the asset distribution under a path of aggregate shocks. Each date supplies one observation $(z_t,B_t,p_t,B_{t+1})$. For each $z$, fit the nodal values separately for $v_t=\log p_t$ and $v_t=B_{t+1}$. Let $\theta$ be the vector of nodal values and $X_t$ the two nonzero interpolation weights at $B_t$, so $X_t\theta$ is the forecast. The update solves
$$
\widetilde\theta_z=\arg\min_\theta\left\{
\sum_{t:z_t=z}(v_t-X_t\theta)^2
+0.002\|\theta-\theta_z^{\mathrm{old}}\|_2^2
+0.001\|D\theta\|_2^2
\right\},\qquad
(D\theta)_j=\theta_j-2\theta_{j+1}+\theta_{j+2}.
$$
The $0.002$ term keeps poorly sampled nodes near the previous rule; the $0.001$ term smooths neighboring nodal values. Only observations from the simulated path enter the squared-error term. Mean assets $B$ stay in levels because they can be negative.

At each outer iteration $k$, report the fit of the **current** perceived rules to that iteration's simulated path. For $v_t=B_{t+1}$ or $v_t=\log p_t$, let $\widehat v_t$ be the corresponding forecast and $\bar v_z$ the mean of $v_t$ among dates with $z_t=z$. The two reported numbers are computed separately:
$$
\boxed{R^2_{v,k}=1-\frac{\sum_t(v_t-\widehat v_t)^2}
{\sum_z\sum_{t:z_t=z}(v_t-\bar v_z)^2}},
\qquad v\in\{B_{t+1},\log p_t\}.
$$

Update both forecasting functions and repeat:
$$
\boxed{
(\widehat P,\widehat H)
\ \longrightarrow\ (g,\lambda)
\ \longrightarrow\ \{F_t,B_t,p_t,B_{t+1}\}_{t=1}^T
\ \longrightarrow\ (\widetilde P,\widetilde H).
}
$$
Use the same aggregate draws across outer rounds. Stop when the fitted rules stabilize on the simulated states.

## 7. Numerical solution and results

Use the grids in Section 6.1. Fit the forecasting rules from 8,000 simulated dates after a 500-date burn-in. Each date contributes its realized aggregate state and outcomes. At outer iteration $k$, compare the current rules with the rules newly fitted to that iteration's simulated path:
$$
\Delta_k=\max_{1\le t\le T}\max\left\{
\left|\log\frac{\widetilde P_k(z_t,B_t)}{\widehat P_k(z_t,B_t)}\right|,
\left|\widetilde H_k(z_t,B_t)-\widehat H_k(z_t,B_t)\right|
\right\}.
$$
The printed `update` is $\Delta_k$, evaluated before damping; it is not the maximum change in the nodal coefficients. At a candidate stopping iteration, households are re-solved at tolerance $2\times10^{-8}$ if needed, and $\Delta_k$ is recomputed. The **outer-loop stopping condition** uses this checked value:
$$
\boxed{\Delta_k<2\times10^{-4}.}
$$
The run is capped at 140 outer iterations; reaching the cap without meeting this condition reports `converged=false`. The solver also prints separate $R^2$ values for $B_{t+1}$ and $\log p_t$ each iteration.

```julia
# One KS-style simulation run with a shared 300-node asset grid.
base_model = HA.model(;nb=300,nf=300,nB=160,Bmin=-1.10,Bmax=-0.30,risk=0.05)
B_nodes = vcat(collect(range(-1.10,-1.02,length=17)),
               collect(range(-1.02,-0.82,length=101))[2:end],
               collect(range(-0.82,-0.30,length=44))[2:end])
m = merge(base_model,(;bf=copy(base_model.b),B=B_nodes,nf=300,nB=length(B_nodes)))
@assert length(m.B)==160 && m.b==m.bf && all(diff(m.B).>0)
result = HA.solve_equilibrium(m;T=8000,burn=500,seed=20260925,
                           maxiter=140,tol=2e-4,verbose=true)
println("Simulation-regression run complete: ", length(result.history),
        " forecast updates; converged=", result.converged)
```

### 7.1 Simulated economy

Simulate the solved household and aggregate rules. The figure shows mean assets, the market-clearing and forecast nontradable prices, the constrained share, and the average asset distribution.

```julia
s = HA.simulate(result.m,result.hh,result.forecast;
             T=3000,burn=500,seed=271828,keep=16)
m = result.m
t = 1:min(500,s.T)
fig = plot(layout=(2,2),size=(1100,720))
plot!(fig[1],t,s.B[t];label="Actual B",ylabel="Mean assets",xlabel="Date")
plot!(fig[2],t,s.price[t];label="Clearing price",ylabel="Nontradable price",xlabel="Date")
plot!(fig[2],t,s.ppred[t];label="Forecast",linestyle=:dash)
plot!(fig[3],t,100 .* s.binding[t];label=false,ylabel="Constrained households (%)",xlabel="Date")
plot!(fig[4],m.bf,cumsum(s.Faverage);label=false,ylabel="Average asset CDF",xlabel="Inherited assets b",
      xlims=(-1.15,-0.35),ylims=(0,1))
savefig(fig,joinpath(FIGDIR,"02_simulation.png")); display(fig)
showtable(["Simulated moment", "Value"], [
    ("Mean net foreign assets",mean(s.B)), ("Std. dev. of mean assets",std(s.B)),
    ("Mean nontradable price",mean(s.price)), ("Mean collateral-constrained share",mean(s.binding)),
    ("Mean cross-sectional asset variance",mean(s.variance)),
])
```

### 7.2 Household policy functions

Plot two fixed aggregate states from the simulation in Section 7.1: the higher-endowment state $z=8$ and the lower-endowment state $z=7$. For each $z$, take its first saved simulation date and hold the inherited distribution $F_z^\star$ fixed. Thus $B_z^\star=\sum_j b_j^F F_{j,z}^\star$ and the market-clearing price $p_z^{N,\star}$ are fixed **within that figure**. Only individual inherited assets $b$ (the horizontal axis) and the two income realizations $\varepsilon$ vary. The code prints $B_z^\star$, $p_z^{N,\star}$, and the constrained household share for each state. In this simulation, the $z=8$ distribution has no constrained households, while some households bind at $z=7$. Dotted horizontal lines show the collateral limits.

```julia
function plot_policy_state(z_policy,filename)
    snapshot_index = findfirst(x -> x.z == z_policy,s.snapshots)
    @assert !isnothing(snapshot_index) "No saved snapshot with z = $z_policy"
    ss = s.snapshots[snapshot_index]
    F_policy = ss.F
    B_policy = dot(m.bf,F_policy)
    @assert isapprox(B_policy,ss.B;atol=1e-10)
    vpol = HA.clear_market(m,result.hh,result.forecast,F_policy,z_policy)
    binding_share = s.binding[ss.k]
    println("Fixed policy state: z = ",z_policy,", B = ",round(B_policy,digits=5),
            ", pN = ",round(vpol.price,digits=5),
            ", constrained share = ",round(100*binding_share,digits=2),
            "%, simulated date = ",ss.k)

    assets = collect(range(-1.10,-0.35,length=220))
    pol = HA.current_policies(m,vpol.M,z_policy,vpol.price,assets)
    fig = plot(layout=(1,2),size=(1100,400))
    policy_colors = [:dodgerblue,:darkorange]
    for e in 1:m.ne
        label = "epsilon = $(m.e[e])"
        plot!(fig[1],assets,pol.g[:,e];label,color=policy_colors[e],
              xlabel="Inherited assets b",ylabel="Next assets b'")
        hline!(fig[1],[HA.collateral(m,z_policy,m.e[e],vpol.price)];
               label=false,color=policy_colors[e],linestyle=:dot)
        plot!(fig[2],assets,pol.cT[:,e];label,color=policy_colors[e],
              xlabel="Inherited assets b",ylabel="Tradable consumption")
    end
    plot!(fig[1],assets,assets;label="45-degree line",color=:gray,linestyle=:dash)
    plot!(fig[1];title="z = $z_policy, B = $(round(B_policy,digits=3))",titlefontsize=10)
    plot!(fig[2];title="Constrained share: $(round(100*binding_share,digits=1))%",titlefontsize=10)
    savefig(fig,joinpath(FIGDIR,filename))
    display(fig)
end

plot_policy_state(8,"03_policies.png")
plot_policy_state(7,"03_policies_z7.png")
```

### 7.3 Estimated aggregate forecasting functions

Each panel fixes one aggregate state $z$. The blue curve is the estimated rule, $\log\widehat P(z,B)$ or $\widehat H(z,B)$. The gray band marks the 1st–99th percentile of simulated $B_t$ conditional on that $z$ in the final solver iteration. Vertical scales differ across panels.

```julia
function plot_aggregate_rule(result, which)
    m = result.m; f = result.forecast; sim = result.simulation
    fig = plot(layout=(4,4),size=(1650,1250),legend=false,
               left_margin=5Plots.mm,bottom_margin=4Plots.mm,
               top_margin=4Plots.mm,right_margin=2Plots.mm)
    for z in 1:m.nz
        visited_B = sim.B[sim.z .== z]
        lo,hi = quantile(visited_B,[0.01,0.99])
        vspan!(fig[z],[lo,hi];color=:lightgray,alpha=0.25,label=false)
        values = which == :price ? f.logP[:,z] : f.H[:,z]
        plot!(fig[z],m.B,values;color=:steelblue,linewidth=2,label=false)
        plot!(fig[z];title="z=$z | yT=$(round(m.yT[z],digits=3)), yN=$(round(m.yN[z],digits=3))",
              titlefontsize=8,xtickfontsize=7,ytickfontsize=7,
              xlims=(first(m.B),last(m.B)),
              xlabel=z>12 ? "Mean assets B" : "",
              ylabel=(z-1)%4==0 ? (which == :price ? "log forecast price" : "Forecast B next") : "")
    end
    fig
end

pricefig = plot_aggregate_rule(result,:price)
bondfig = plot_aggregate_rule(result,:bond)
savefig(pricefig,joinpath(FIGDIR,"04_forecast_log_price.png"))
savefig(bondfig,joinpath(FIGDIR,"05_forecast_next_assets.png"))
display(pricefig)
display(bondfig)
```
