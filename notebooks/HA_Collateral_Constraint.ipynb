# Heterogeneous Agents in a Small Open Economy with Collateral Borrowing Constraints

- Authors: Yuxuan Zhao
- Date: 2026-09-23

This notebook defines a heterogeneous-household extension of the endowment economy in [Bianchi (2011)](../reference/2011-overborrowing-and-systemic-externalities-in-the-business-cycle.pdf)

Households differ in tradable endowments and bond holdings. The aggregate shock follows a Markov chain; idiosyncratic income shocks are IID. The full asset distribution is part of the aggregate state, and the joint asset-income distribution factors into the asset distribution and the fixed IID income distribution. The exposition proceeds from economic fundamentals to the recursive household problem and then to the recursive competitive equilibrium. All equilibrium objects are defined using the full distribution.

## 1. Economic fundamentals

### 1.1 Households, goods, and financial markets

Time is discrete, $t=0,1,2,\ldots$. There is a unit mass of infinitely lived, price-taking households indexed by $i\in[0,1]$. Households have common preferences and face uninsurable idiosyncratic income risk as well as aggregate risk.

There are two consumption goods. Tradables can be exchanged internationally and their price is normalized to one. Nontradables are exchanged within the domestic economy at the common relative price $p_t^N>0$.

Each household can trade a one-period, non-state-contingent bond with the rest of the world at the exogenous gross return
$$
R=1+r>1.
$$
The bond is denominated in tradable goods. The inherited position is $b_{i,t}$ and the position chosen at date $t$ is $b_{i,t+1}$. Positive holdings represent assets; negative holdings represent debt. Households cannot trade state-contingent claims. Both sectors are endowment sectors.

### 1.2 Aggregate and idiosyncratic endowments

Let $z_t\in\mathcal Z$ be a finite-state aggregate Markov shock, with
$$
\Pr(z_{t+1}=z'\mid z_t=z)=\Pi(z'\mid z),
\qquad \sum_{z'\in\mathcal Z}\Pi(z'\mid z)=1.
$$
The common endowment components are functions $y_c^T(z)$ and $y^N(z)>0$. Household $i$ has an idiosyncratic tradable-income component $\varepsilon_{i,t}\in\mathcal E$, where $\mathcal E$ is a finite set. Its endowments are
$$
\boxed{
y_{i,t}^T=y^T(z_t,\varepsilon_{i,t})
=y_c^T(z_t)+\varepsilon_{i,t},
\qquad y_{i,t}^N=y^N(z_t).
}
$$
Choose the endowment functions and shock supports so that
$$
y^T(z,\varepsilon)>0
\qquad\text{for all admissible }(z,\varepsilon).
$$

Idiosyncratic shocks are independent and identically distributed across households and time, and independent of the entire aggregate shock process:
$$
\boxed{\varepsilon_{i,t}\overset{\mathrm{iid}}{\sim}\pi_\varepsilon,}
\qquad
\pi_\varepsilon(\varepsilon)\geq0,\quad
\sum_{\varepsilon\in\mathcal E}\pi_\varepsilon(\varepsilon)=1,\quad
\sum_{\varepsilon\in\mathcal E}\varepsilon\pi_\varepsilon(\varepsilon)=0.
$$
Thus the household's joint transition probabilities are
$$
\Pr(z',\varepsilon'\mid z,\varepsilon)
=\Pi(z'\mid z)\pi_\varepsilon(\varepsilon').
$$
Current and past individual income shocks do not predict the next idiosyncratic draw. IID shocks still allow runs of favorable or adverse realizations; past shocks can have lasting effects through accumulated assets and debt.

Assume a conditional exact law of large numbers for the continuum of households. Conditional on the aggregate history, the idiosyncratic cross-sectional distribution is $\pi_\varepsilon$, including within groups defined by inherited assets. Initial assets are independent of the initial idiosyncratic draw. Consequently,
$$
\int_0^1\varepsilon_{i,t}\,di=0,
\qquad
Y_t^T=\int_0^1 y_{i,t}^T\,di=y_c^T(z_t).
$$

### 1.3 Preferences

Each household maximizes
$$
\mathbb E_0\sum_{t=0}^{\infty}\beta^t u(C_{i,t}),
\qquad 0<\beta<1,
$$
where
$$
u(C)=
\begin{cases}
\dfrac{C^{1-\sigma}}{1-\sigma}, & \sigma>0,\ \sigma\ne1,\\
\log C, & \sigma=1,
\end{cases}
$$
and the consumption basket is
$$
C_{i,t}=C(c_{i,t}^T,c_{i,t}^N)
=\left[\omega(c_{i,t}^T)^{-\eta}
+(1-\omega)(c_{i,t}^N)^{-\eta}\right]^{-1/\eta},
\qquad 0<\omega<1,\quad\eta>-1,\quad\eta\ne0.
$$
At $\eta=0$, use the continuous Cobb-Douglas limit
$$
C(c^T,c^N)=(c^T)^\omega(c^N)^{1-\omega}.
$$
The elasticity of substitution between tradable and nontradable consumption is $1/(1+\eta)$. Both consumption goods are strictly positive.

### 1.4 Budget and collateral borrowing constraints

For every household and date,
$$
\boxed{
b_{i,t+1}+c_{i,t}^T+p_t^Nc_{i,t}^N
=Rb_{i,t}+y^T(z_t,\varepsilon_{i,t})+p_t^Ny^N(z_t).
}
$$
Borrowing is limited by the market value of the household's current endowments:
$$
\boxed{
b_{i,t+1}\geq
-\left[\kappa^Ty^T(z_t,\varepsilon_{i,t})
+\kappa^Np_t^Ny^N(z_t)\right],
\qquad 0\leq\kappa^T,\kappa^N<1.
}
$$
The parameters $\kappa^T$ and $\kappa^N$ are pledgeable fractions. A negative idiosyncratic income shock reduces both current resources and, when $\kappa^T>0$, borrowing capacity. A lower common price $p_t^N$ reduces every household's nontradable collateral value when $\kappa^N>0$.

Households take prices as given. They may buy or sell nontradables domestically, so individual consumption need not equal individual endowment:
$$
c_{i,t}^N\ne y^N(z_t)\quad\text{in general}.
$$
Admissible plans are adapted to observed histories, satisfy both constraints with positive consumption at every history of positive probability, have well-defined expected lifetime utility, and obey the no-Ponzi restriction
$$
\liminf_{T\to\infty}R^{-T}b_{i,T}\geq0
\qquad\text{almost surely}.
$$
There is no default or additional transfer instrument. Equilibrium is defined on states admitting feasible continuation plans.

## 2. States, distribution, and timing

### 2.1 Individual and aggregate states

Let $F_t$ be the cross-sectional probability distribution of inherited bond holdings and let $\Phi_t$ denote the joint distribution of inherited assets and current idiosyncratic income:
$$
F_t(A)=\int_0^1\mathbf 1\{b_{i,t}\in A\}\,di,
\qquad
\Phi_t(A\times E)=\int_0^1
\mathbf 1\{b_{i,t}\in A,\ \varepsilon_{i,t}\in E\}\,di,
$$
for measurable $A\subseteq\mathbb R$ and $E\subseteq\mathcal E$.

Assets $b_{i,t}$ are chosen before the fresh IID draw $\varepsilon_{i,t}$. Conditional on the aggregate history, their joint distribution therefore factors as
$$
\boxed{
\Phi_t(A\times E)=F_t(A)\sum_{\varepsilon\in E}\pi_\varepsilon(\varepsilon),
\qquad \Phi_t=F_t\otimes\pi_\varepsilon.
}
$$
It is sufficient to track the full asset distribution $F_t$, with
$$
\int_{\mathbb R}dF_t(b)=1,
\qquad \int_{\mathbb R}|b|\,dF_t(b)<\infty.
$$
The household's own state is $(b,\varepsilon)$, the aggregate state is $(z,F)$, and the full decision state is
$$
\boxed{s=(b,\varepsilon,z,F).}
$$
Current $\varepsilon$ remains a household state because it affects current resources and the collateral limit. Inherited assets retain the effects of past shocks. Assets chosen today, $b_{i,t+1}$, generally depend on today's $\varepsilon_{i,t}$; the independence statement concerns $b_{i,t}$ and $\varepsilon_{i,t}$.

Aggregate net foreign assets and endowments are
$$
B(F)=\int b\,dF(b),
\qquad
Y^T(z)=\sum_{\varepsilon\in\mathcal E}\pi_\varepsilon(\varepsilon)
\int y^T(z,\varepsilon)\,dF(b)=y_c^T(z),
\qquad Y^N(z)=y^N(z).
$$
Here and below, asset integrals are over $\mathbb R$. The factorization is exact under the IID assumptions; it does not replace the asset distribution with its mean $B$. Different asset distributions can generate different consumption and borrowing decisions even when their means coincide.

### 2.2 Within-period timing and aggregate laws

1. Households enter date $t$ with inherited assets distributed according to $F_t$. Current shocks $z_t$ and $\varepsilon_{i,t}$ are observed. The joint asset-income distribution is $\Phi_t=F_t\otimes\pi_\varepsilon$.
2. Given the aggregate state, households trade nontradables at $p_t^N=p^N(z_t,F_t)$ and choose $(c_{i,t}^T,c_{i,t}^N,b_{i,t+1})$ simultaneously.
3. These saving choices determine the next inherited-asset distribution $F_{t+1}$. Next period's aggregate shock is drawn from $\Pi$ and fresh idiosyncratic shocks are drawn independently from $\pi_\varepsilon$. The next joint distribution is $\Phi_{t+1}=F_{t+1}\otimes\pi_\varepsilon$.

Given a perceived asset-distribution law $\mathcal H$, households forecast
$$
\boxed{F'=\mathcal H(F,z),}
\qquad p^{N\prime}=p^N(z',F').
$$
The distribution $F'$ is determined before $z'$ is realized, while next-period prices and choices depend on the realized $z'$. Households observe the aggregate state and their own states and know $\Pi$ and $\pi_\varepsilon$. Each household has measure zero, so its own deviation in $b'$ does not change $F'$ or the price functions it faces.

## 3. Recursive household problem

### 3.1 Bellman equation

Given the common price function $p^N$ and perceived law of motion $\mathcal H$, the household solves
$$
\boxed{
\begin{aligned}
V(b,\varepsilon;z,F)
=\max_{b',\,c^T>0,\,c^N>0}\ \Bigg\{&
u\!\left(C(c^T,c^N)\right)\\
&+\beta\sum_{z'\in\mathcal Z}\Pi(z'\mid z)
\sum_{\varepsilon'\in\mathcal E}\pi_\varepsilon(\varepsilon')
V\!\left(b',\varepsilon';z',\mathcal H(F,z)\right)
\Bigg\},
\end{aligned}
}
$$
subject to
$$
\begin{aligned}
b'+c^T+p^N(z,F)c^N
&=Rb+y^T(z,\varepsilon)+p^N(z,F)y^N(z),\\
b'&\geq\underline b(\varepsilon;z,F)
\equiv-\left[\kappa^Ty^T(z,\varepsilon)
+\kappa^Np^N(z,F)y^N(z)\right],\\
F'&=\mathcal H(F,z).
\end{aligned}
$$
The maximization is over choices admitting an admissible continuation plan as defined above. Its policy functions are
$$
b'=g(b,\varepsilon;z,F),\qquad
c^T=c^T(b,\varepsilon;z,F),\qquad
c^N=c^N(b,\varepsilon;z,F).
$$
The borrowing bound depends on the household's current income and the common equilibrium price. The expectation includes the Markov aggregate shock and a fresh IID income draw. The next asset distribution $F'=\mathcal H(F,z)$ is already determined by current decisions; next-period prices also depend on $z'$.

### 3.2 Household optimality conditions

Define marginal utilities with respect to the two goods by
$$
u_T(c^T,c^N)=u'(C(c^T,c^N))C_T(c^T,c^N),
\qquad
u_N(c^T,c^N)=u'(C(c^T,c^N))C_N(c^T,c^N).
$$
Let $\lambda(s)$ be the budget multiplier and $\mu(s)\geq0$ the collateral multiplier. Under the usual differentiability conditions, optimal policies satisfy
$$
\lambda(s)=u_T(c^T(s),c^N(s)),
\qquad
p^N(z,F)\lambda(s)=u_N(c^T(s),c^N(s)).
$$
In particular, optimal intratemporal allocation requires
$$
p^N(z,F)=\frac{1-\omega}{\omega}
\left(\frac{c^T(s)}{c^N(s)}\right)^{1+\eta}.
$$
For a transition $(z',\varepsilon')$, write
$$
s'=\left(g(s),\varepsilon',z',\mathcal H(F,z)\right).
$$
The Euler and complementary-slackness conditions are
$$
\lambda(s)=\beta R
\sum_{z'\in\mathcal Z}\Pi(z'\mid z)
\sum_{\varepsilon'\in\mathcal E}\pi_\varepsilon(\varepsilon')
\lambda(s')+\mu(s),
$$
$$
\begin{aligned}
\Delta(s)&=g(s)+\kappa^Ty^T(z,\varepsilon)
+\kappa^Np^N(z,F)y^N(z)\geq0,\\
\mu(s)&\geq0,\qquad \mu(s)\Delta(s)=0.
\end{aligned}
$$
These are private optimality conditions: households do not differentiate the equilibrium price with respect to their own choices. Lifetime optimality is specified by the household problem, rather than by these local conditions alone.

## 4. Recursive competitive equilibrium

Given the primitives
$$
\left\{\beta,\sigma,\omega,\eta,R,\kappa^T,\kappa^N,
\mathcal Z,\mathcal E,\Pi,\pi_\varepsilon,y_c^T,y^N\right\}
$$
and an admissible initial state $(z_0,F_0)$ with $\Phi_0=F_0\otimes\pi_\varepsilon$, a recursive competitive equilibrium is a collection
$$
\boxed{\left\{V,\ g,\ c^T,\ c^N,\ p^N,\ \mathcal H\right\}}
$$
and its induced distribution process
$$
F_{t+1}=\mathcal H(F_t,z_t),
\qquad z_{t+1}\sim\Pi(\cdot\mid z_t),
$$
satisfying the following conditions at every reachable aggregate state $(z,F)$.

### 4.1 Household optimization

Given $p^N$ and $\mathcal H$, $V$ is the value function of the recursive household problem. For every admissible individual state, the policy triple attains its maximum:
$$
\begin{aligned}
&\left(g(s),c^T(s),c^N(s)\right)\\
&\quad\in\operatorname*{arg\,max}_{b',\,x^T>0,\,x^N>0}
\left\{u(C(x^T,x^N))+
\beta\sum_{z',\varepsilon'}\Pi(z'\mid z)\pi_\varepsilon(\varepsilon')
V(b',\varepsilon';z',\mathcal H(F,z))\right\},
\end{aligned}
$$
subject to the budget, collateral, and continuation-feasibility restrictions in Section 3.1. In particular,
$$
g(s)+c^T(s)+p^N(z,F)c^N(s)
=Rb+y^T(z,\varepsilon)+p^N(z,F)y^N(z),
\qquad g(s)\geq\underline b(\varepsilon;z,F).
$$

### 4.2 Consistency of the distribution law

Given the saving policy $g$ and IID income probabilities $\pi_\varepsilon$, the actual asset-distribution law $\mathcal G_g$ is defined, for every measurable asset set $A\subseteq\mathbb R$, by
$$
\boxed{
\mathcal G_g(F,z)(A)
=\sum_{\varepsilon\in\mathcal E}\pi_\varepsilon(\varepsilon)
\int\mathbf 1\!\left\{g(b,\varepsilon;z,F)\in A\right\}\,dF(b).
}
$$
Rational expectations require the perceived and actual laws to coincide as probability measures:
$$
\boxed{\mathcal H(F,z)=\mathcal G_g(F,z).}
$$
Fresh IID income draws then imply, for every $\varepsilon'\in\mathcal E$,
$$
\Phi'(A\times\{\varepsilon'\})
=\mathcal H(F,z)(A)\pi_\varepsilon(\varepsilon').
$$
The sum in the asset update averages over current shocks $\varepsilon$, which affect today's saving decisions. The next income draw $\varepsilon'$ is independent of the assets already selected. Both distribution statements hold conditional on any next aggregate realization $z'$ with $\Pi(z'\mid z)>0$.

### 4.3 Nontradable goods market clearing

At the common price $p^N(z,F)>0$, aggregate nontradable demand equals the domestic endowment:
$$
\boxed{
\sum_{\varepsilon\in\mathcal E}\pi_\varepsilon(\varepsilon)
\int c^N(b,\varepsilon;z,F)\,dF(b)=y^N(z).
}
$$
Define aggregate consumption of good $j\in\{T,N\}$ by
$$
C^j(z,F)=\sum_{\varepsilon\in\mathcal E}\pi_\varepsilon(\varepsilon)
\int c^j(b,\varepsilon;z,F)\,dF(b).
$$
Here $C^T$ and $C^N$ denote aggregate quantities of the separate goods; $C(c^T,c^N)$ denotes an individual CES basket.

### 4.4 Bond aggregation and the tradable resource constraint

Current and next-period aggregate net foreign assets satisfy
$$
B=B(F)=\int b\,dF(b),
\qquad
B'=\sum_{\varepsilon\in\mathcal E}\pi_\varepsilon(\varepsilon)
\int g(b,\varepsilon;z,F)\,dF(b).
$$
Distribution consistency also implies
$$
B\!\left(\mathcal H(F,z)\right)=B'.
$$
Both $F'$ and its mean $B'$ are determined by choices made before $z'$ is realized.

With mean-zero IID income shocks, the aggregate tradable resource constraint is
$$
\boxed{C^T(z,F)+B'=RB(F)+y_c^T(z).}
$$
This condition follows by aggregating household budgets and imposing nontradable market clearing. Foreign lenders accommodate the net bond position at the world return $R$; domestic net bond holdings are not required to sum to zero.

### 4.5 An implied aggregate price relation

With common CES preferences and a common price, household optimality implies
$$
c^N(b,\varepsilon;z,F)
=q\!\left(p^N(z,F)\right)c^T(b,\varepsilon;z,F),
\qquad
q(p)=\left(\frac{1-\omega}{\omega p}\right)^{1/(1+\eta)}.
$$
Aggregating and using nontradable market clearing gives
$$
q\!\left(p^N(z,F)\right)C^T(z,F)=y^N(z),
$$
or equivalently,
$$
\boxed{
p^N(z,F)=\frac{1-\omega}{\omega}
\left(\frac{C^T(z,F)}{y^N(z)}\right)^{1+\eta}.
}
$$
This is a consequence of household optimality and market clearing, not an additional independent equilibrium condition. It does not impose $c_i^N=y^N$ household by household. The full distribution affects aggregate consumption through individual saving and borrowing decisions, which in turn determine the common relative price and collateral values.

## 5. Calibration and economic inputs

Sections 1–4 define the exact heterogeneous-household economy. We now specify its numerical inputs and construct an approximate decentralized equilibrium. The only distribution statistic used in household forecasts will be mean assets, $B$. Simulation will still track the entire asset distribution.

One period is **one year**. Tradables are the numeraire, and the stationary mean of aggregate tradable endowment is one. Thus a bond position of $-1$ is debt equal to one year's average tradable endowment. Household mass is one. The world interest rate is exogenous; aggregate bond holdings are the economy's net foreign assets and need not sum to zero.

### 5.1 Preferences, interest rate, and collateral

The following are the **exact values used in the computation**. Calibration targets and their rationale come from [Bianchi (2011), Section IV.A and Table 1, pp. 3410–3412](../reference/2011-overborrowing-and-systemic-externalities-in-the-business-cycle.pdf#page=11). The unrounded parameter values come from the author's [replication code, `main.m`, parameter block](../reference/MatlabCode_overborrowingAER_web/main.m). The paper reports some values at lower precision.

| Parameter | Value used | Meaning and source / calibration target |
| --- | --- | --- |
| σ | 2 | CRRA curvature. Standard benchmark in the original annual small-open-economy calibration. |
| β | 0.906 | Annual discount factor. Original target: average net foreign assets relative to GDP of −29% in Argentina; the paper uses Lane–Milesi-Ferretti data. |
| ω | 0.3070 | Tradable weight in the CES consumption aggregator. Original target: tradable production accounts for 32% of the value of output, using Argentine sectoral data. The CES weight itself is not that observed output share. |
| 1/(1+η) | 0.83 | Intratemporal elasticity between tradables and nontradables. The paper chooses the upper end of the empirical range 0.40–0.83, a benchmark with weaker price amplification. |
| η | 1/0.83−1=0.2048192771… | Derived CES curvature; not a separately calibrated parameter. |
| r | 0.04 | Annual net world risk-free interest rate; standard value in the original calibration. |
| R=1+r | 1.04 | Derived gross bond return. |
| κᵀ | 0.3235 | Tradable collateral coefficient. Original target: sudden-stop frequency of 5.5%, from the cross-country evidence cited by the paper. |
| κᴺ/κᵀ | 1 | Original baseline assumption: equal collateral coefficients for the two endowments. |
| κᴺ | 0.3235 | Follows from κᴺ=κᵀ. |

In the original representative-household model, $(\beta,\omega,\kappa^T)$ are chosen jointly to match the three long-run targets. The sudden-stop target refers to an episode with a binding credit constraint and a sufficiently large increase in net capital outflows, not simply the fraction of constrained households. Here these parameters are **inherited calibration inputs**: after introducing household heterogeneity, we have not re-estimated them or re-matched those three targets. The HA moments computed below therefore need not equal the original target moments.

### 5.2 Aggregate endowment risk: a 16-state Markov chain

The common state $z_t\in\{1,\ldots,16\}$ indexes the pair $(y_c^T(z_t),y^N(z_t))$. Its transition matrix is
$$
\Pi_{zz'}=\Pr(z_{t+1}=z'\mid z_t=z),\qquad \sum_{z'=1}^{16}\Pi_{zz'}=1.
$$
Both sectoral endowments fluctuate in this inherited aggregate process. Idiosyncratic risk, introduced below, affects only the household's tradable endowment.

The paper estimates a bivariate VAR from HP-filtered log cyclical components of Argentine tradable and nontradable GDP, using WDI annual data for 1965–2007. Manufacturing and primary products are classified as tradables; the remaining sectors are nontradables. Using $\nu_t$ for the **aggregate innovation**, the estimated process is
$$
\begin{pmatrix}\log y_{c,t}^T\\ \log y_t^N\end{pmatrix}
=A\begin{pmatrix}\log y_{c,t-1}^T\\ \log y_{t-1}^N\end{pmatrix}+\nu_t,
\qquad \nu_t\sim N(0,\Sigma),
$$
$$
A=\begin{pmatrix}0.901&0.495\\-0.453&0.225\end{pmatrix},
\qquad
\Sigma=\begin{pmatrix}0.00219&0.00162\\0.00162&0.00167\end{pmatrix}.
$$
The paper uses a Tauchen–Hussey discretization with four points per shock dimension and 16 joint states. We use the **supplied discrete chain** from [`proc_shock.mat`](../reference/MatlabCode_overborrowingAER_web/proc_shock.mat), as exported in [`shocks.jl`](../src/bianchi2011/shocks.jl), rather than reconstructing it from the rounded VAR coefficients above. Rows of $\Pi$ are normalized to remove floating-point row-sum error. The next output lists all endowment pairs and the stationary probabilities $\pi_z$, which solve
$$
\pi_{z'}=\sum_z\pi_z\Pi_{zz'},\qquad \sum_z\pi_z=1.
$$
The full transition matrix is printed in Appendix B, in the same state order. The actual chain satisfies $\sum_z\pi_z y_c^T(z)=\sum_z\pi_z y^N(z)=1$ to floating-point precision.

### 5.3 Idiosyncratic risk and other normalizations

| Input | Value used | Interpretation / source |
| --- | --- | --- |
| Household mass | 1 | Normalization, so integrals of assets and consumption are per-household aggregates. |
| Tradable price | 1 | Numeraire. The endogenous pᴺ is a relative price, not an additional calibrated parameter. |
| Mean aggregate endowments | E[yᵀ_c] = E[yᴺ] = 1 | Original normalization; verified on the supplied Markov chain. |
| Idiosyncratic support | ε ∈ {−0.05, +0.05} | New illustrative assumption, not estimated from household data. This is an additive shock of five percent of mean aggregate tradable endowment. |
| Idiosyncratic probabilities | (1/2,1/2) | Equal probabilities; E[ε]=0, sd(ε) = 0.05. |
| Idiosyncratic persistence | 0 | IID across dates and households, independent of z. Each row of its two-state transition matrix is (1/2,1/2). |

Thus $y_{i,t}^T=y_c^T(z_t)+\varepsilon_{i,t}$ and $y_{i,t}^N=y^N(z_t)$. All tradable endowments are positive. Although income draws are IID, assets retain the effect of past income realizations. At the beginning of a period, the joint distribution is $F_t(b)\pi_\varepsilon(\varepsilon)$ because current income is independent of inherited assets. The code integrates over the two income realizations exactly; it does not draw a finite sample of households.

## 6. What is approximated, and what is stored on grids?

### 6.1 Household forecasts and the simulated economy

The exact household problem has aggregate state $(z,F)$. We approximate the information about $F$ used for forecasting by its mean,
$$
B=m(F)=\int b\,dF(b).
$$
At outer iteration $n$, households take two functions as given:
$$
\boxed{p=\widehat P_n(z,B),\qquad \widehat B'=\widehat H_n(z,B).}
$$
The first predicts today's nontradable price; the second predicts next period's mean inherited assets. Saving is chosen before $z'$ is realized, so $\widehat H_n$ has no $z'$ argument. The perceived next price is $\widehat P_n(z',\widehat H_n(z,B))$.

For these rules, the approximate household problem is
$$
\widehat V_n(b,\varepsilon;z,B)
=\max_{\substack{b'\in\mathbb R\\c^T>0,\ c^N>0}}\left\{
u(C(c^T,c^N))+
\beta\sum_{z',\varepsilon'}\Pi_{zz'}\pi_\varepsilon(\varepsilon')
\widehat V_n(b',\varepsilon';z',\widehat H_n(z,B))\right\},
$$
subject to the budget and collateral constraints of Section 3, with $p=\widehat P_n(z,B)$. Bond holdings $b'$ may be negative. We compute the policy using the Euler conditions rather than iterating directly on $\widehat V_n$.

Simulation maintains the whole histogram $F_t$. Given $F_t$, it solves for a price $p_t$ that clears the nontradable market, allowing $p_t$ to differ from $\widehat P_n(z_t,B_t)$. Households adjust their current choices to that price, while their continuation expectations still use $\widehat H_n$ and the household solution under the perceived rules. The resulting actual $F_{t+1}$ and $B_{t+1}$ generate observations for updating both forecasts.

This construction lets the approximation make observable mistakes. Setting $p_t$ equal to its forecast or forcing $B_{t+1}=\widehat H_n(z_t,B_t)$ during simulation would suppress precisely the forecast errors we need to measure.

### 6.2 Three distinct asset grids

We need three grids because policy interpolation, distribution transport, and aggregate forecasting are different operations.

| Grid | Symbol and domain | What is stored / evaluated there |
| --- | --- | --- |
| Individual policy grid | Individual b grid: [−1.35, 1.00] | Household saving and marginal utility as functions of current assets b. Next assets are continuous choices within the numerical domain, not restricted to these nodes. |
| Distribution grid | Histogram b grid: [−1.35, 1.00] | Probability masses Fℓ,t at inherited assets bℓᶠ. It is finer so mass can move with less discretization error. |
| Aggregate mean grid | Mean B grid: [−1.10, −0.30] | Forecasts of the common price and next mean assets, and the aggregate asset argument of household policies. It is a grid for B, not an asset constraint on individual households. |

The three experiments use:

| Run | `n_b` | Individual-grid construction | `n_F` | Distribution spacing | `n_B` | Aggregate-grid construction |
| --- | --- | --- | --- | --- | --- | --- |
| Baseline | 121 | 97 nodes on [−1.35,−0.45], then 24 additional nodes on (−0.45,1] | 601 | 2.35/600=0.00391667 | 41 | Uniform on [−1.10,−0.30]; spacing 0.02 |
| Refined | 241 | 193 nodes on [−1.35,−0.45], then 48 additional nodes on (−0.45,1] | 1001 | 2.35/1000=0.00235 | 81 | Uniform on the same interval; spacing 0.01 |
| Coverage | 241 | Same as refined | 1001 | Same as refined | 161 | Union of the refined grid and 101 points on [−1.02,−0.82], with duplicates removed; spacing 0.002 in that interval and 0.01 outside it |

Each of the two individual-grid segments is evenly spaced; the common point $-0.45$ appears once. Spacings are $0.009375$ and $0.06041667$ in the baseline, and $0.0046875$ and $0.03020833$ in the refinement. This places more nodes in the debt region where collateral constraints create kinks, while retaining positive saving choices. The distribution grid is uniform throughout. Coverage-grid nodes are rounded to 12 decimal places before duplicates are removed.

These domains and node counts are numerical choices for this HA experiment, not parameters calibrated to Argentine data. The lower policy endpoint must cover the economic borrowing limit, the upper saving endpoint must be inactive, and simulated means must remain inside the aggregate grid. Those requirements are checked below. They do not establish that the chosen resolution is already sufficient.

### 6.3 Interpolation and policy arrays

For adjacent nodes $x_j\le x\le x_{j+1}$ define $w=(x-x_j)/(x_{j+1}-x_j)$. Linear interpolation is
$$
\mathcal I[v](x)=(1-w)v_j+w v_{j+1}.
$$
For example, a query $B=-0.905$ between nodes $-0.91$ and $-0.90$ uses weights $(1/2,1/2)$. Aggregate shocks and IID income states are summed over exactly; they are not interpolated.

For each $z$, store forecast coefficients $a_{j,z}=\log\widehat P(z,B_j)$ and $h_{j,z}=\widehat H(z,B_j)$. At an intermediate $B$,
$$
\log\widehat P(z,B)=\sum_j\varphi_j(B)a_{j,z},\qquad
\widehat H(z,B)=\sum_j\varphi_j(B)h_{j,z},
$$
where $\varphi_j$ are the linear weights just defined. Interpolating log prices guarantees positive prices. Mean assets are represented in **levels**, since $B$ is negative in this experiment.

The household arrays $g(b_i,\varepsilon_e;z,B_j)$ and $\lambda(b_i,\varepsilon_e;z,B_j)$ have $n_b\times2\times n_B\times16$ entries. Continuation values use linear interpolation in each of $b$ and $B$ (bilinear interpolation), without interpolation across discrete shocks. Asset interpolation outside its domain raises an error, apart from a $10^{-9}$ endpoint roundoff allowance. Forecast functions have constant extensions at the ends of the $B$ grid; actual simulated $B_t$ and its one-step forecast are separately required to lie within that grid.

### 6.4 The order of computation

| Calculation | Inputs held fixed | Object updated | Section |
| --- | --- | --- | --- |
| Household time iteration | Price and asset rules at outer round n, economic parameters and grids | Policy g and marginal utility λ | 7 |
| Current-price search | Fₜ, zₜ and the household continuation marginal-value function | Trial price p, and current choices at that price | 8.1–8.2 |
| Distribution transition | Clearing price and current household choices | Full Fₜ₊₁ and its actual mean | 8.3 |
| Aggregate-rule iteration | Same aggregate random draws and numerical configuration | Forecast coefficients (a,h) | 9 |

The policy iteration index is $k$, the outer forecast iteration index is $n$, and $t$ denotes an economic date in the simulation. A household solve is completed before starting that outer round's simulation.

## 7. Step 1: solve households given the two aggregate rules

### 7.1 Reduce the two consumption choices at a given price

Fix $(b,\varepsilon,z,B)$ and let $p=\widehat P_n(z,B)$. The household takes this price as given. Its intratemporal condition is
$$
\frac{u_N(c^T,c^N)}{u_T(c^T,c^N)}=p
\quad\Longrightarrow\quad
c^N=q(p)c^T,\qquad
q(p)=\left(\frac{1-\omega}{\omega p}\right)^{1/(1+\eta)}.
$$
Thus we can recover both consumption goods from expenditure after choosing $b'$. Define resources $x$ and the economic lower bound $\ell$ by
$$
x=Rb+y_c^T(z)+\varepsilon+p y^N(z),\qquad
\ell=-\kappa^T[y_c^T(z)+\varepsilon]-\kappa^N p y^N(z).
$$
Define the following price-dependent quantities:
$$
\begin{aligned}
d(p)&=1+p q(p),&
A_c(p)&=[\omega+(1-\omega)q(p)^{-\eta}]^{-1/\eta},\\
c^T&=\frac{x-b'}{d(p)},&
c^N&=q(p)c^T,\qquad C=A_c(p)c^T.
\end{aligned}
$$
The budget multiplier is marginal utility of tradable expenditure:
$$
\lambda=u_T=\omega C^{1+\eta-\sigma}(c^T)^{-1-\eta}
=K(p)(c^T)^{-\sigma},\qquad
K(p)=\omega A_c(p)^{1+\eta-\sigma}.
$$
These formulas solve the within-period allocation analytically. Individual nontradable consumption remains a choice and generally differs from individual nontradable endowment. Only **aggregate** nontradable demand will be set equal to aggregate supply in Step 2.

### 7.2 Initialize a feasible saving policy

Hold the aggregate rules fixed throughout this household solve. If there is no previous solution on this grid, initialize saving at $b'=b$ and clip it to
$$
g^{(0)}=\operatorname{clip}\left(b,\ell,\min\{b_{\max},x-10^{-8}\}\right),
\qquad b_{\max}=1.
$$
Here $\operatorname{clip}(v,L,U)=\min\{\max\{v,L\},U\}$. The small expenditure margin ensures positive consumption. In later outer rounds on the same grid, use the previous household policy instead of $b$, then apply the same feasibility clipping because the price forecast may have changed.

### 7.3 Compute expected future marginal utility from the old policy

Given the entire table $g^{(k)}$, first compute $\lambda^{(k)}$ at every household grid state using Section 7.1. At a current aggregate state $(z,B)$ let $\widehat B'=\widehat H_n(z,B)$. For any candidate next asset position $a$, define
$$
\mathcal M^{(k)}(a;z,B)
=\beta R\sum_{z'=1}^{16}\sum_{e'=1}^{2}
\Pi_{zz'}\pi_{\varepsilon,e'}\,
\mathcal I_{b,B}[\lambda^{(k)}](a,\varepsilon_{e'};z',\widehat B').
$$
The future marginal utilities already incorporate the future price forecasts and the future saving choices implied by $g^{(k)}$. IID income implies that $\pi_{\varepsilon,e'}=1/2$ does not depend on today's income. The transition of $z$, by contrast, depends on today's $z$.

In implementation, first interpolate in $B$ at $\widehat B'$ and sum over $(z',e')$ for each individual asset node $a=b_i$. This creates a one-dimensional table of $\mathcal M^{(k)}$. Interpolate that table at the **old current saving choice**:
$$
M^{(k)}=\mathcal M^{(k)}(g^{(k)}(b,\varepsilon;z,B);z,B).
$$
Freezing this right-hand side gives one explicit update. It is not a one-pass solution to the Euler equation: $\mathcal M^{(k)}$ and its saving argument are updated repeatedly until the policy stops changing.

### 7.4 Update consumption and saving, with the borrowing constraint

At an interior solution the Euler condition is $\lambda=\beta R E[\lambda']$. At the collateral limit, the condition becomes $\lambda\ge\beta R E[\lambda']$: the household would like to transfer more resources from the future but cannot borrow more.

For the current update, compute consumption when the household borrows to its limit and its associated marginal utility:
$$
c^T_{\rm bound}=\frac{x-\ell}{d(p)},\qquad
\lambda_{\rm bound}=K(p)(c^T_{\rm bound})^{-\sigma}.
$$
There are two economic cases:

1. **Borrowing-bound proposal.** If $\lambda_{\rm bound}\ge M^{(k)}$, set $\widetilde c^T=c^T_{\rm bound}$ and $\widetilde g=\ell$. Even at maximum feasible current expenditure, current marginal utility is at least its discounted expected future value.
2. **Interior proposal.** If $\lambda_{\rm bound}<M^{(k)}$, invert $K(p)(c^T)^{-\sigma}=M^{(k)}$:
$$
c^T_* =\left(\frac{K(p)}{M^{(k)}}\right)^{1/\sigma},\qquad
\widetilde g=x-d(p)c^T_*.
$$
This uses less current expenditure and chooses more assets than the borrowing limit.

The finite grid also has an upper saving endpoint. Set $u=\min\{b_{\max},x-10^{-12}\}$ and $c^T_{\min}=(x-u)/d(p)$. All cases can then be implemented as
$$
\widetilde c^T=\operatorname{clip}(c^T_*,c^T_{\min},c^T_{\rm bound}),
\qquad \widetilde g=x-d(p)\widetilde c^T.
$$
The upper endpoint is a numerical truncation, not an economic saving constraint. We report whether it binds; if it does, the asset domain needs to be enlarged before interpreting the result.

### 7.5 Dampen the proposal and repeat

After computing proposals at every grid state, update
$$
g^{(k+1)}=0.75g^{(k)}+0.25\widetilde g^{(k)}.
$$
Return to Section 7.3 unless the **undamped** maximum policy change satisfies
$$
\max_{i,e,j,z}|\widetilde g^{(k)}-g^{(k)}|<\tau_{HH}.
$$
The tight tolerance is $\tau_{HH}=2\times10^{-8}$ in asset units; the adaptive choice in intermediate outer rounds is specified in Section 9.4. The maximum is 6,000 household iterations, after which failure to converge raises an error. Recompute the marginal-utility table from the returned damped policy so that $g$ and $\lambda$ are consistent.

This is fixed-grid Euler time iteration. The CRRA–CES formula makes the scalar consumption inversion analytic; it does not remove either good from the household optimization problem.

### 7.6 A numerical example of the two update branches

The next cell illustrates **one update**, before running the household solver. Fix $b=-0.9$, the low income realization, the aggregate state closest to mean endowments, and a current price from the initial rule. Compare two hypothetical continuation values: $M=0.8\lambda_{\rm bound}$ and $M=1.2\lambda_{\rm bound}$. The first should propose the borrowing limit; the second should propose interior saving. These two $M$ values are teaching examples, not additional calibration parameters or equilibrium estimates.

```julia
f0 = initial_forecast(m0)
let z = argmin(abs.(m0.yT.-1).+abs.(m0.yN.-1)), b = -0.9, eps = m0.e[1]
    price = HA.price_forecast(f0,z,mean(m0.B))
    st = HA.consumption_at_price(price,m0.p)
    cash = HA.resources(m0,b,eps,z,price)
    lower = HA.collateral(m0,z,eps,price)
    cbound = (cash-lower)/st.d
    lambound = st.K*cbound^(-m0.p.sigma)
    rows = map((0.8,1.2)) do factor
        M = factor*lambound
        cstar = (st.K/M)^(1/m0.p.sigma)
        proposal = HA.current_choice(m0,cash,lower,st,M)
        cT = (cash-proposal)/st.d
        @assert abs(proposal+cT+price*st.q*cT-cash) < 1e-12
        @assert factor < 1 ? abs(proposal-lower)<1e-12 : proposal>lower
        (factor,lower,cstar,cbound,cT,proposal,proposal-lower)
    end
    showtable(["M / lambda_bound", "Borrowing limit", "Unconstrained cT", "cT at limit",
               "Chosen cT", "Proposed b'", "Collateral slack"],rows)
end
```

## 8. Step 2: simulate market clearing and the full distribution

### 8.1 Fix the continuation object for the current date

After Step 1, hold the converged household table $\lambda$ and aggregate rules fixed. At date $t$, the inputs are the current aggregate shock $z_t$ and masses $F_{\ell,t}$ on the distribution grid. Compute
$$
B_t=\sum_\ell b_\ell^F F_{\ell,t},\qquad
\widehat B_{t+1}=\widehat H_n(z_t,B_t),
$$
and construct
$$
\mathcal M_t(a)=\beta R\sum_{z',e'}\Pi_{z_tz'}\pi_{\varepsilon,e'}
\mathcal I_{b,B}[\lambda](a,\varepsilon_{e'};z',\widehat B_{t+1}).
$$
Keep this **whole function of $a$** fixed while trying different current prices. Its value at a household's chosen $a=b'$ can still change as that household changes saving. Today's trial price does not replace the next-period price rule inside this continuation object.

### 8.2 At each trial price, recompute choices and aggregate demand

Start the price search near $\widehat P_n(z_t,B_t)$. At a trial $p$:

1. For every occupied distribution node $b_\ell^F$ and both current income states, recompute $x$, $\ell$, $q(p)$, $d(p)$ and $K(p)$. A price change changes both the value of nontradable income and the collateral limit.
2. Solve today's household choices by repeating the update in Section 7.4 with $M=\mathcal M_t(g^{(j)})$. Here $j$ is a local policy iteration and the continuation function stays fixed. Use damping $0.45$, tolerance $5\times10^{-9}$ on the undamped saving proposal, and at most 1,200 iterations. The first price evaluation starts from feasible clipped $b'=b$; subsequent evaluations start from the previous trial's choices and clip them to the new feasible set.
3. Recover $c_t^T$ and $c_t^N=q(p)c_t^T$, then calculate nontradable excess demand:
$$
\mathcal R_N(p;F_t,z_t)
=\sum_{\ell,e}F_{\ell,t}\pi_{\varepsilon,e}
c_t^N(b_\ell^F,\varepsilon_e;p)-y^N(z_t).
$$
4. Use a sign-changing price bracket and Brent's method to find $p_t$ satisfying $\mathcal R_N(p_t;F_t,z_t)=0$. The absolute **price** tolerance is $2\times10^{-9}$; goods-market residuals are checked separately. Appendix A specifies the bracket and feasibility bounds.

All positive distribution masses enter demand; none is dropped during simulation. Every trial price entails a new current-household solve. Simply evaluating the policy table computed at $\widehat P_n(z_t,B_t)$ would not give household choices consistent with a different clearing price $p_t$.

### 8.3 Move mass to next period's asset distribution

Let $g_t(b_\ell^F,\varepsilon_e;p_t)$ be the saving choice at the clearing price. It usually lies between two distribution nodes. If $b_j^F\le g_t\le b_{j+1}^F$, assign the corresponding household mass using
$$
w_j(g_t)=\frac{b_{j+1}^F-g_t}{b_{j+1}^F-b_j^F},\qquad
w_{j+1}(g_t)=\frac{g_t-b_j^F}{b_{j+1}^F-b_j^F}.
$$
All other weights are zero. For example, if the neighboring nodes are $-0.902$ and $-0.900$ and saving is $-0.9015$, weights are $0.75$ and $0.25$. A mass of $0.01$ contributes $0.0075$ and $0.0025$ to those nodes; its mean saving is preserved exactly. These example nodes illustrate the calculation, not a replacement for the grids in Section 6.

Sum these contributions over current assets and current income:
$$
\boxed{F_{j,t+1}=\sum_{\ell,e}F_{\ell,t}\pi_{\varepsilon,e}
w_j\!\left(g_t(b_\ell^F,\varepsilon_e;p_t)\right).}
$$
This two-node lottery preserves total probability and aggregate saving, up to floating-point error:
$$
\sum_jF_{j,t+1}=1,\qquad
B_{t+1}=\sum_j b_j^F F_{j,t+1}
=\sum_{\ell,e}F_{\ell,t}\pi_{\varepsilon,e}g_t(b_\ell^F,\varepsilon_e;p_t).
$$
The new IID income draw gives $\Phi_{t+1}=F_{t+1}\otimes\pi_\varepsilon$. In particular, no persistence matrix for individual income needs to be carried through the asset transition. Combining household budgets with nontradable clearing gives the aggregate tradable resource identity
$$
C_t^T+B_{t+1}=RB_t+y_c^T(z_t).
$$
Record the differences between the actual outcome and the two forecasts:
$$
e_t^B=B_{t+1}-\widehat H_n(z_t,B_t),\qquad
e_t^p=\frac{p_t}{\widehat P_n(z_t,B_t)}-1.
$$
These errors will usually be nonzero. Carry the actual $F_{t+1}$ into the next date rather than replacing it with a distribution having the forecast mean.

The next cell executes one date under the initial rules, using a two-node initial histogram with mean $B=-0.8$ for this standalone example. It reports prices, actual and forecast next assets, and accounting residuals. The complete training simulations use the initial distributions specified in Section 8.4.

```julia
F0 = initial_distribution(m0; B=-0.8)
z0 = argmin(abs.(m0.yT .- 1) .+ abs.(m0.yN .- 1))
one_period = HA.clear_market(m0, hh0, f0, F0, z0)
showtable(["One-period market/distribution check", "Value"], [
    ("Initial mean B", one_period.B), ("Clearing price", one_period.price),
    ("Predicted next mean", one_period.Bpred), ("Actual next mean", one_period.Bnext),
    ("Nontradable excess demand", one_period.market_error),
    ("Tradable resource residual", one_period.resource_error),
    ("Next probability mass minus one", sum(one_period.Fnext)-1),
])
```

### 8.4 Generate the simulation sample

Use a total of 8,500 annual dates: discard the first 500 as burn-in and retain 8,000 for forecast fitting. Construct the aggregate path with Julia's `Xoshiro` generator and seed `20260925`. The initial state is the supplied state closest to $(y_c^T,y^N)=(1,1)$ in absolute distance (state 10 in this chain). Subsequent states are drawn from the corresponding row of $\Pi$.

At the beginning of every outer round, restart from the same initial distribution and the same aggregate draws for that run. The initial histogram is the two-node lottery with mean equal to the arithmetic mean of that run's aggregate-grid nodes: $-0.7$ in the baseline and refined runs, and approximately $-0.80931677$ in the nonuniform coverage run. It is concentrated at at most two adjacent distribution nodes, not uniform over the asset domain. Using a fixed path removes changes in aggregate random draws as a source of noise in the forecast update. IID household income is always integrated exactly.

For each retained date, save $(z_t,B_t,p_t,B_{t+1})$, market and resource residuals, the constrained share, and distribution moments. These observations are generated with the complete histogram even though the forecast regressions use only $(z_t,B_t)$.

## 9. Step 3: fit and iterate the aggregate forecasting rules

### 9.1 Fit one price rule and one asset rule for each aggregate state

The unknowns are the coefficients $a_{j,z}$ and $h_{j,z}$ defined in Section 6.3. For a fixed $z$, collect the fitting observations with that aggregate state. In row $t$ of the design matrix $X_z$, put the linear weights $\varphi_j(B_t)$. Each row has at most two nonzero entries that sum to one.

Fit the price rule to $v_t=\log p_t$ and the asset rule to $v_t=B_{t+1}$. For either coefficient vector $\theta_z\in\{a_{\cdot,z},h_{\cdot,z}\}$, the proposal is the regularized least-squares solution
$$
\widetilde\theta_z=\arg\min_\theta\left\{
\|v_z-X_z\theta\|_2^2
+\rho\|\theta-\theta_z^n\|_2^2
+\chi\|D_2\theta\|_2^2\right\},
\quad \rho=0.002,\quad\chi=0.001.
$$
Here $(D_2\theta)_j=\theta_j-2\theta_{j+1}+\theta_{j+2}$. The first penalty keeps unsupported nodes close to the previous rule; the second discourages rough changes across consecutive node coefficients. For the nonuniform grid this is still an unscaled coefficient second-difference penalty, not a finite-difference approximation to curvature in physical $B$ units. Both penalty strengths are numerical choices, not calibrated economic parameters.

Compute the proposal by the linear system
$$
\boxed{(X_z'X_z+0.002 I+0.001 D_2'D_2)\widetilde\theta_z
=X_z'v_z+0.002\theta_z^n.}
$$
The loss is a sum over observations, not a sample average. Every observation has unit weight. Require at least 12 observations for each aggregate state. After solving, restrict proposed $h$ nodes to $[-1.10+10^{-5},-0.30-10^{-5}]$ and log-price nodes to the feasibility intervals in Appendix A. These restrictions keep future queries and the household grid feasible; their use on the simulated path is diagnosed separately.

### 9.2 Optional additional observations for rare state combinations

The third run uses the denser 161-node aggregate grid and augments the fitting sample as follows:

1. In its first outer simulation, select the retained date whose $B_t$ is nearest each aggregate-grid node. Also select dates $80,160,\ldots,8000$. Remove duplicate dates.
2. Freeze these **date indices** for later outer rounds. Their full histograms are regenerated as the rules change; freezing dates prevents changes in the selection rule itself from moving the regression weights around.
3. For every selected histogram $F_t$, solve the current market under each of the 16 possible aggregate shocks using Step 2. Each solution supplies another $(z,B,p,B')$ row with unit regression weight. All aggregate transitions in the supplied chain have positive probability.
4. Append these rows to the 8,000 ordinary simulation rows and apply Section 9.1 unchanged.

This makes rarely visited $(z,B)$ combinations count more in fitting. It does not add a new household state, replace the histogram by a point mass, or add fictitious observations to reported economic moments. The baseline and refined runs use only their ordinary paths. Comparisons with the third run therefore change both aggregate-grid resolution and the fitting measure; they cannot attribute an improvement to just one of those changes.

### 9.3 Measure the proposed update, then damp it

Before damping, compare the old and proposed forecasts on the set $\mathcal S_n$ of fitting states:
$$
d_n=\max_{(z,B)\in\mathcal S_n}\max\left\{
\left|\log\frac{\widetilde P_n(z,B)}{\widehat P_n(z,B)}\right|,
\left|\widetilde H_n(z,B)-\widehat H_n(z,B)\right|
\right\}.
$$
In the third run, $\mathcal S_n$ includes the extra rows. Also report the same update norm over every aggregate-grid node and shock as the **whole-grid update**. The latter is a diagnostic, not the stopping criterion; unsupported grid regions may settle more slowly.

If another outer round is needed, update both coefficient tables by
$$
a^{n+1}=0.75a^n+0.25\widetilde a^n,\qquad
h^{n+1}=0.75h^n+0.25\widetilde h^n.
$$
Price damping is in log prices, while asset damping is in levels. Return to Step 1 with these new rules. Warm-start household policies from the preceding round on the same grid. The undamped norm $d_n$ avoids making convergence appear better merely by choosing a smaller damping factor.

### 9.4 Household precision and the stopping rule

Use tight household precision in the first round. Thereafter let
$$
\tau_{HH}^{n}=
\begin{cases}
2\times10^{-8},&n=0,\\
\operatorname{clip}(0.001d_{n-1},2\times10^{-8},2\times10^{-5}),&n\ge1.
\end{cases}
$$
This permits a looser household solve when the aggregate rules are still moving substantially. If $d_n<2\times10^{-4}$, or the outer iteration cap is reached, and households were solved more loosely, **resolve households at $2\times10^{-8}$, resimulate, refit, and recompute $d_n$** before deciding whether to stop.

Declare convergence only if the rechecked undamped update satisfies
$$
\boxed{d_n<2\times10^{-4}.}
$$
The returned candidate is the current rule and household solution against which the small proposed change was measured. Reaching the iteration cap alone returns `converged = false`. The caps are 140, 100, and 65 outer rounds for the baseline, refined, and coverage runs, respectively.

This stopping rule describes a fixed point of the numerical projection on its fitting states. It is neither a bound on price forecast errors nor a proof that $B$ contains all relevant distribution information. Section 11 tests the resulting policies against actual future market clearing.

### 9.5 Algorithm in execution order

1. Load economic parameters and the Markov chain; construct the three grids.
2. Initialize $(\widehat P_0,\widehat H_0)$ as in Section 7.7, or interpolate a previous run's aggregate rules onto the new grid.
3. Solve households for the current rules (Section 7).
4. Restart the fixed aggregate path and initial histogram; at each date clear the market, move mass, and store the actual outcomes (Section 8).
5. For the coverage run only, add the selected histograms evaluated under all aggregate shocks (Section 9.2).
6. Fit the two rule proposals and compute their undamped update norm (Sections 9.1 and 9.3).
7. If stopping is proposed, perform the tight-precision recheck. Otherwise damp the rules and return to step 3.
8. On a separate aggregate path, check forecasts, budgets, markets, distribution mass, boundaries, and independent Euler inequalities (Section 11).

## 11. Validate convergence and equilibrium accuracy separately

### 11.1 A held-out path and explicit forecast-error measures

Development checks used seed `9711` to locate coverage problems. After choosing the numerical design, use seed `271828` for the validation reported here: 3,000 retained dates after 500 burn-in dates. This path is not used to fit coefficients or select the extra training dates.

For this simulation define
$$
e_t^B=B_{t+1}-\widehat H(z_t,B_t),\qquad
e_t^p=p_t/\widehat P(z_t,B_t)-1.
$$
For either error series, $\operatorname{RMSE}(e)=\sqrt{T^{-1}\sum_t e_t^2}$ and the maximum error is $\max_t|e_t|$. Asset errors are in tradable units; price errors are relative to the forecast and are multiplied by 100 in percentage tables. Also report
$$
R_B^2=1-\frac{\sum_t(e_t^B)^2}{\sum_t(B_{t+1}-\overline{B'})^2}.
$$
To expose accumulated forecast drift, start $\widetilde B_1=B_1$ and recursively compute $\widetilde B_{t+1}=\widehat H(z_t,\widetilde B_t)$ along the same aggregate shocks, without resetting to the observed mean. Report $\max_t|\widetilde B_{t+1}-B_{t+1}|$. A small one-step error or high $R^2$ alone cannot guarantee small recursive error.

### 11.2 Accounting, constraints, and numerical-domain checks

Check both the household table under perceived rules and the simulated choices. The Euler residual under perceived continuation values is $1-\mathcal M(g)/\lambda$. Interior choices require zero; choices at the borrowing limit require a nonnegative residual. Classify collateral binding using saving slack below $10^{-7}$. The table-grid saving ceiling is classified with the same distance threshold.

On the full validation path, check nontradable excess demand, the aggregate tradable resource identity, distribution mass, and mass at distribution endpoints. At retained distributions, additionally check household budgets, minimum collateral slack, and the perceived-rule Euler equation. The latter household checks skip nodes with asset mass at most $10^{-12}$ for reporting only. All masses still enter market clearing and distribution transport.

The assertions below require market/resource residuals below $10^{-7}$, household budget/mass errors below $10^{-10}$, collateral slack above $-10^{-9}$, endpoint mass below $10^{-10}$, and zero table states at the numerical saving ceiling. They also require no out-of-grid aggregate forecast at household grid states and no on-path contact with the forecast guards defined in Appendix A. Passing these implementation checks does not establish accuracy of the distribution approximation.

### 11.3 Re-clear future markets for an independent Euler check

Retain 16 regularly spaced validation distributions (dates $187,374,\ldots,2992$) and also the dates with the lowest price, highest constrained share, largest price forecast error, and largest asset forecast error; remove duplicate dates. At each retained $(F,z)$:

1. Re-clear today's market, obtain each household's $b'$, and construct the actual next distribution $F'$.
2. For every possible next aggregate shock $z'$, clear the nontradable market at $(F',z')$. This gives an actual next price, using the actual next distribution rather than the forecast next mean.
3. Evaluate each current household's chosen $b'$ at that next price for each fresh IID income realization. These asset positions need not lie on the policy or histogram grids. Solve their next-period current choices with the fitted continuation rules and calculate their marginal utilities.
4. Integrate these marginal utilities over $(z',\varepsilon')$ and report
$$
\mathcal E(b,\varepsilon;z,F)
=1-\frac{\beta R\sum_{z',\varepsilon'}\Pi_{zz'}\pi_\varepsilon(\varepsilon')
\lambda_{\rm recleared}(b',\varepsilon';z',F')}{\lambda(b,\varepsilon;z,F)}.
$$

For interior households report $\max|\mathcal E|$ and the mean absolute residual weighted by household masses across sampled dates. For households at the collateral bound report $\max\{0,-\mathcal E\}$: a positive residual at that bound is consistent with the Euler inequality and is not itself an error. Exclude asset nodes with mass at most $10^{-8}$ from these independent summary statistics, but retain them in simulation.

Unlike the solver's own Euler residual, this calculation replaces forecast next means and prices with actual distribution transitions and future-market clearing. It measures the combined consequences of distribution compression and numerical approximation. Next-period decisions still use the fitted continuation rules, so this is a diagnostic of the approximate policy, not a separate exact solution of the full-distribution model.

## Appendix A. Feasibility guards and price brackets

This appendix records the small numerical safeguards used by the algorithm. They are part of the implementation and should be inspected if the policy domain, calibration, or income support is changed.

### A.1 Admissible forecast-price nodes

Write $y_e^T(z)=y_c^T(z)+\varepsilon_e$ and $b_{\min}=-1.35$. Positive expenditure at the borrowing limit, even at the lowest policy-grid asset, requires
$$
Rb_{\min}+(1+\kappa^T)y_e^T(z)+(1+\kappa^N)p y^N(z)>0.
$$
Requiring that the grid contain the collateral lower bound gives $-\kappa^T y_e^T(z)-\kappa^N p y^N(z)\ge b_{\min}$. Consequently the fitted price nodes are clipped to $[p_L(z),p_U(z)]$, with
$$
p_L(z)=\max\left\{0.01,\ \max_e
\frac{-Rb_{\min}-(1+\kappa^T)y_e^T(z)}{(1+\kappa^N)y^N(z)}+0.02\right\},
$$
$$
p_U(z)=\min_e\frac{-b_{\min}-\kappa^T y_e^T(z)}{\kappa^N y^N(z)}-0.02.
$$
The clipping is applied to the corresponding log-price coefficients. Asset forecast nodes are clipped to $[B_{\min}+10^{-5},B_{\max}-10^{-5}]$. Linear interpolation between admissible nodes preserves these bounds. If economically relevant forecasts repeatedly touch the guards, the grid/parameter domain needs attention; the guards should not be interpreted as restrictions of the exact model.

For the reported on-path guard share, examine interpolation nodes with weight greater than $10^{-10}$. Count a date as touching a guard if a relevant log-price coefficient is within $10^{-6}$ of either log-price bound, or a next-asset coefficient is within $2\times10^{-5}$ of a $B$-domain endpoint.

### A.2 Trial-price bracket in simulation

For the **occupied** current assets, define the feasibility floor and the grid-coverage ceiling
$$
p_{\rm floor}=\max_{\ell:F_\ell>0,\ e}
\frac{-Rb_\ell^F-(1+\kappa^T)y_e^T(z)}{(1+\kappa^N)y^N(z)},\qquad
p_{\rm ceil}=\min_e\frac{-b_{\min}-\kappa^T y_e^T(z)}{\kappa^N y^N(z)}.
$$
With $p_0=\widehat P(z,B)$, start from
$$
L=\max\{0.8p_0,p_{\rm floor}+0.01,0.02\},\qquad
U=\min\{1.2p_0,p_{\rm ceil}-0.001\}.
$$
Evaluate excess demand at both endpoints. If they do not bracket a root, expand up to 12 times using
$$
L\leftarrow\max\{0.8L,p_{\rm floor}+0.001,0.005\},\qquad
U\leftarrow\min\{1.2U,p_{\rm ceil}-0.0001\}.
$$
Failure to obtain opposite signs raises an error. A valid bracket is passed to Brent's method with absolute price tolerance $2\times10^{-9}$ and zero relative tolerance. The implementation checks positive feasible resources at occupied nodes, enforces the borrowing bound and individual-grid coverage, and reports the final market residual. A bracketed root is the price selected by this algorithm; these operations by themselves do not prove global uniqueness of equilibrium.

### A.3 Initial feasibility and endpoint margins

Initial policies leave at least $10^{-8}$ of expenditure; subsequent policy proposals use the positive-expenditure margin $10^{-12}$. Local current-household solves check $x>\ell+10^{-10}$. The collateral limit may be below the lowest policy node only within $10^{-10}$ of floating-point roundoff; interpolation accepts endpoints within $10^{-9}$. None of these small constants substitutes for enlarging a grid whose economic domain is inadequate.
